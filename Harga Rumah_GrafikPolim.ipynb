## **Algoritma**


---



1.   Input Data
  *   membaca data `"harga_rumah.csv"` yang telah diinput sebelumnya pada program
  

2.   Inisialisasi Variabel
  *   hitung jumlah data `n`
  *   ambil data dari kolom `luas` dan `harga` menjadi array `x` dan `y`


3. Menghitung Nilai Agregat
  *   menghitung jumlah :
    *   Σx  : jumlah luas
    *   Σy  : jumlah harga
    *   Σxy : jumlah perkalian luas dan harga
    *   Σx^2  : jumlah luas kuadrat
    *   Σx^3  : jumlah luas pangkat tiga
    *   Σx^4  : jumlah luas pangkat empat
    *   Σx^2 y : jumlah luas kuadrat dikali harga


4. Susun Sistem Persamaan Linear
    *   buat matriks koefisien `A` untuk persamaan
    ​
\begin{bmatrix}
\Sigma x^4 & \Sigma x^3 & \Sigma x^2 \\
\Sigma x^3 & \Sigma x^2 & \Sigma x \\
\Sigma x^2 & \Sigma x   & n
\end{bmatrix}
"""

    *   buat matriks hasil `B`

     \begin{bmatrix}
\Sigma x^2 y \\
\Sigma xy \\
\Sigma y
\end{bmatrix}


5. Penyelesaian Sistem Linear
    *   Gunakan metode `np.linalg.solve()` atau eliminasi Gauss untuk menyelesaikan sistem persamaan linear dan mendapatkan nilai koefisien `a, b, dan c` yang memenuhi persamaan polinomial.
      *   `A . koefesien = B`
      *   hasil nilai `a, b, c` untuk fungsi polinomial : `y=ax^2+bx+c`


6. Output Koefisien
    * menampilkan nilai koefisien `a, b, c` yang diperoleh dari hasil perhitungan sistem persamaan


7. Pembuatan Grafik
    * Tentukan rentang nilai `x` untuk menggambar grafik fungsi polinomial, dengan `np.linspace`.
    * Hitung nilai y berdasarkan fungsi polinomial `y = ax² + bx + c` menggunakan koefisien yang diperoleh.


8. Visualisasi Hasil
    * Buat grafik dengan menampilkan data asli (`scatter plot`) dan kurva polinomial hasil perhitungan.
    * Berikan label pada sumbu `x` dan `y` serta judul grafik yang sesuai.
    * Tampilkan grid dan legend pada grafik.








## **Kode Program**


---



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('harga_rumah.csv')

x = df['luas']
y = df['harga']
n = len(x)
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x2 = np.sum(x**2)
sum_x3 = np.sum(x**3)
sum_x4 = np.sum(x**4)
sum_x2y = np.sum((x**2) * y)


A = np.array([
    [sum_x4, sum_x3, sum_x2],
    [sum_x3, sum_x2, sum_x],
    [sum_x2, sum_x,  n]
])


B = np.array([
    [sum_x2y],
    [sum_xy],
    [sum_y]
])


coefficients = np.linalg.solve(A, B)
a_value, b_value, c_value = coefficients.flatten()


print("Nilai a:", a_value)
print("Nilai b:", b_value)
print("Nilai c:", c_value)


x_range = np.linspace(min(x), max(x), 1000)
y_range = a_value * x_range**2 + b_value * x_range + c_value


plt.scatter(x, y, color='red', label='Data Asli')
plt.plot(x_range, y_range, color='blue', label='Fungsi Polinomial')
plt.title(f'Grafik Polinomial: $y = {a_value:.2f}x^2 + {b_value:.2f}x + {c_value:.2f}$')
plt.xlabel('luas')
plt.ylabel('harga')
plt.grid()
plt.show()
```

```python
def prediksi_harga(luas):
    harga = a_value * luas**2 + b_value * luas + c_value
    return harga

# Contoh penggunaan: prediksi harga untuk luas 7420
luas_rumah = 7420
harga_prediksi = prediksi_harga(luas_rumah)
print(f"Harga rumah untuk luas {luas_rumah} adalah Rp {harga_prediksi:,.0f}")
```
